## Practice Exercise 1.1: Sales Performance by Region

::: {.callout-tip}
## Your AI Collaboration Challenge

**Business Question**: "Which region is performing best, and what's the average sale size by region?"

**Your Task**: Practice prompting AI with this business context. Try creating a prompt that asks for:

**My Prompt for 1.1:**
```
Consider that I am a Business Analyst working for an office supply company. I have been provided with a dataset named sales_data.csv, which contains the following columns:

Date: Sale date
Product: Name of the product sold
Quantity: Number of units sold
Unit_Price: Price per unit (including $ sign)
Total_Sale: Total value of each transaction
Sales_Rep: Employee responsible for the sale
Region: Geographic region of the sale

Using this dataset, I need to develop a Python program using pandas to analyze sales performance and generate meaningful business insights. 

The program should:
1. Calculate total sales by region
2. Calculate average sale amount by region
3. Calculate the number of transactions by region
4. Format the results in a professional table suitable for a manager presentation
5. Include a short business quenstion answer identifying the best-performing region and what's the average sale size by region
Make sure the code is clean, beginner-friendly, and well-commented.
```

**AI Response:**

````python

import pandas as pd

# Read sales dataset
df = pd.read_csv("sales_data.csv")

# Clean Total_Sale column by removing symbols
df["Total_Sale"] = pd.to_numeric(
    df["Total_Sale"].str.replace("$", "", regex=False).str.replace(",", ""),
    errors="coerce"
)

# Create regional summary table
region_summary = (
    df.groupby("Region")["Total_Sale"]
    .agg(["sum", "mean", "count"])
    .reset_index()
)

# Rename columns
region_summary.columns = [
    "Region",
    "Total_Sales",
    "Average_Sale_Size",
    "Number_of_Transactions"
]

# Round values
region_summary["Average_Sale_Size"] = (
    region_summary["Average_Sale_Size"].round(2)
)

# Format currency columns
region_summary["Total_Sales"] = region_summary["Total_Sales"].apply(
    lambda x: f"${x:,.2f}"
)

region_summary["Average_Sale_Size"] = region_summary["Average_Sale_Size"].apply(
    lambda x: f"${x:,.2f}"
)

# Print formatted table
print("\nSales Performance by Region")
print("=" * 55)
print(region_summary)

# Find top region
top_region = (
    df.groupby("Region")["Total_Sale"]
    .sum()
    .sort_values(ascending=False)
    .index[0]
)

# Display business insights
print("\nBusiness Insight:")
print(f"{top_region} is the strongest performing region based on total revenue generated.")

print("\nRegional Average Sale Values")
print("-" * 35)

for _, row in region_summary.iterrows():
    print(f"{row['Region']} -> {row['Average_Sale_Size']}")
```
````

## Practice Exercise 1.2: Top Performing Products

::: {.callout-tip}
## Business Scenario

Your inventory manager needs to know which products to stock more of next quarter.

**Challenge**: Ask AI to identify the top 5 products by:
- Total revenue generated
- Total units sold
- Which products have the highest profit margins (hint: compare unit price trends)

**Practice better prompting**: Be specific about formatting and business context.
:::

**My Prompt for 1.2:**
```
Create a beginner-level pandas program to examine the sales_data.csv dataset and help the business understand which products are performing the best. 

The program should:
1. Find the top 5 products based on total sales revenue
2. Identify the products with the highest quantities sold
3. Compare products with higher average unit prices to estimate stronger profit opportunities

Display all results in a neat and professional layout using headings and separators for better readability. 
You may include a few simple emojis to make the report more engaging. 

At the end, provide a short and clear business recommendation explaining:
- which products should receive more inventory attention
- whether high-priced or high-volume products are performing better
- what management should focus on for the next quarter

Keep the code simple, readable, and suitable for someone still learning pandas.

**AI Response:**

````python
import pandas as pd

# =========================================
# Import Sales Dataset
# =========================================
df = pd.read_csv("sales_data.csv")

# =========================================
# Prepare Numeric Columns
# =========================================
df["Unit_Price"] = (
    df["Unit_Price"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "")
    .astype(float)
)

df["Total_Sale"] = (
    df["Total_Sale"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "")
    .astype(float)
)

# =========================================
# Highest Revenue Products
# =========================================
revenue_report = (
    df.groupby("Product", as_index=False)["Total_Sale"]
    .sum()
    .sort_values(by="Total_Sale", ascending=False)
    .head(5)
)

# =========================================
# Highest Quantity Sold
# =========================================
quantity_report = (
    df.groupby("Product", as_index=False)["Quantity"]
    .sum()
    .sort_values(by="Quantity", ascending=False)
    .head(5)
)

# =========================================
# Highest Average Unit Price
# =========================================
price_report = (
    df.groupby("Product", as_index=False)["Unit_Price"]
    .mean()
    .sort_values(by="Unit_Price", ascending=False)
    .head(5)
)

# Currency Formatting
revenue_report["Total_Sale"] = revenue_report["Total_Sale"].apply(
    lambda x: f"${x:,.2f}"
)

price_report["Unit_Price"] = price_report["Unit_Price"].apply(
    lambda x: f"${x:,.2f}"
)

# =========================================
# Display Reports
# =========================================
print("\n📈 TOP 5 PRODUCTS BY REVENUE")
print("=" * 65)
print(revenue_report.to_string(index=False))

print("\n📦 TOP 5 PRODUCTS BY QUANTITY SOLD")
print("=" * 65)
print(quantity_report.to_string(index=False))

print("\n💰 TOP 5 PRODUCTS BY AVERAGE UNIT PRICE")
print("=" * 65)
print(price_report.to_string(index=False))

# =========================================
# Business Recommendations
# =========================================
print("\n📋 BUSINESS RECOMMENDATIONS")
print("-" * 65)
print("• Premium technology products generate strong revenue.")
print("• Lower-cost office accessories sell in larger volumes.")
print("• Management should balance inventory between high-value and high-demand products.")
print("• Popular products should receive additional stock next quarter.")
````

**Business Insight Questions :**

- The best performing region is NSW based on total sales.
- High-value technology products generate the most revenue.
- Lower-cost office products sell in larger quantities.
- NSW contributes the highest company revenue.
- The business should maintain stock for high-demand products.

## Practice Exercise 2.2: Budget Category Analysis

::: {.callout-tip}
## Personal Finance Scenario

Someone wants to set a realistic monthly budget based on their spending history.

**Challenge**: Use AI to help:
1. Calculate average monthly spending by category
2. Identify categories with high variation (inconsistent spending)
3. Suggest realistic budget amounts for each category
4. Find categories where spending might be reduced

**Advanced**: Ask AI to explain the statistical concepts it uses.
:::

**Your Budget Analysis:**

````python
import pandas as pd

# =========================
# Load Data
# =========================
transactions = pd.read_csv("transactions.csv")

# =========================
# Standardize Columns
# =========================
transactions.columns = transactions.columns.str.strip().str.capitalize()

# Ensure correct column names
transactions = transactions.rename(columns={
    "Amount": "Amount",
    "Category": "Category"
})

# =========================
# Clean Amount Values
# =========================
transactions["Amount"] = transactions["Amount"].astype(str)
transactions["Amount"] = transactions["Amount"].str.replace("$", "", regex=False)
transactions["Amount"] = transactions["Amount"].str.replace(",", "", regex=False)
transactions["Amount"] = pd.to_numeric(transactions["Amount"], errors="coerce")

# Remove invalid rows
transactions = transactions.dropna(subset=["Amount", "Category"])

# =========================
# ANALYSIS
# =========================
summary = transactions.groupby("Category")["Amount"].agg(
    Total="sum",
    Average="mean",
    Variation="std"
)

# Fix missing variation values
summary["Variation"] = summary["Variation"].fillna(0)

# Suggested budget (10% buffer)
summary["Suggested_Budget"] = summary["Average"] * 1.10

# =========================
# OUTPUT
# =========================
print("\n===== AVERAGE SPENDING BY CATEGORY =====")
print(summary["Average"].to_string())

print("\n===== SPENDING VARIATION BY CATEGORY =====")
print(summary["Variation"].to_string())

print("\n===== TOTAL SPENDING BY CATEGORY =====")
print(summary["Total"].to_string())

print("\n===== FULL BUDGET SUMMARY =====")
print(summary.to_string())

print("\n===== SUGGESTED MONTHLY BUDGET =====")
print(summary[["Suggested_Budget"]].to_string())

# =========================
# INSIGHTS
# =========================
print("\n===== BUDGET INSIGHTS =====")
print("• Focus on categories with higher average spending.")
print("• High variation indicates unstable spending habits.")
print("• Suggested budget includes a 10% safety buffer.")
print("• Reduce unnecessary expenses to improve savings.")
````

### Critique Exercise 1: Overly Complex Code

```python
# AI generated this code - what's wrong with it?
import pandas as pd
import numpy as np
from datetime import datetime

def complex_analysis(data):
    result = data.groupby([pd.Grouper(key='Date', freq='M'), 'Category']).agg({
        'Amount': [lambda x: pd.to_numeric(x.str.replace('$', ''), errors='coerce').sum(),
                   lambda x: pd.to_numeric(x.str.replace('$', ''), errors='coerce').mean(),
                   lambda x: pd.to_numeric(x.str.replace('$', ''), errors='coerce').std()]
    }).reset_index()
    
    result.columns = ['_'.join(col).strip() for col in result.columns.values]
    return result.fillna(0).round(2)

# Complex usage
df['Date'] = pd.to_datetime(df['Date'])
complex_result = complex_analysis(transactions_df)
print(complex_result.head())
```

**My Critique:**

* The code is overly complex due to the use of deeply nested lambda functions within the `agg()` method, which reduces readability and makes the code difficult to maintain.

* Data cleaning operations, such as removing the `$` symbol and converting values using `pd.to_numeric()`, are repeated in multiple places, resulting in inefficiency and unnecessary redundancy.

* The output structure is not clearly presented due to multi-level column names created through automatic renaming (e.g., using `'_'.join()`), making the final results harder to interpret.

* Several processes, including data cleaning, grouping, and summarisation, are combined into a single block of code, which reduces clarity and makes the logic harder to follow.

* The overall implementation is not beginner-friendly and may be challenging for non-technical stakeholders or business users to understand and interpret effectively.
