```python
import pandas as pd
import pyinputplus as pyip
import os
```

Here i imported pandas for data habdling, pyinputplus for input validity, and os to check if the file exists.

```python
def load_data():
    try:
        df = pd.read_csv("budget.csv")
        print("Data loaded successfully.")
    except FileNotFoundError:
        df = pd.DataFrame(columns=['date', 'description', 'type', 'amount'])
        print("No existing file found. Starting fresh.")
    return df
```

In this cell i collected my inputs for date, description, type and all.

```python
def show_menu():
    print("\n1. Add Transaction")
    print("2. View All Transactions")
    print("3. View Balance")
    print("4. Save & Exit")
```

```python
def add_transaction(df):
    date = pyip.inputStr("Enter date (YYYY-MM-DD): ")
    description = pyip.inputStr("Enter description: ")

    t_type = pyip.inputChoice(['income', 'expense'])

    amount = pyip.inputFloat("Enter amount: ", min=0.01)

    new_row = {
        "date": date,
        "description": description,
        "type": t_type,
        "amount": amount
    }

    df.loc[len(df)] = new_row

    print("Transaction added!")

    return df
```

```python
def view_transactions(df):
    if df.empty:
        print("No transactions yet.")
    else:
        print(df)
```

```python
def calculate_balance(df):
    income = df[df['type'] == 'income']['amount'].sum()
    expense = df[df['type'] == 'expense']['amount'].sum()
    return income - expense
```

```python
def save_data(df):
    df.to_csv("budget.csv", index=False)
    print("Data saved successfully.")
```

```python
def main():
    df = load_data()

    while True:
        show_menu()

        choice = pyip.inputInt("Choose option (1-4): ", min=1, max=4)

        if choice == 1:
            df = add_transaction(df)

        elif choice == 2:
            view_transactions(df)

        elif choice == 3:
            balance = calculate_balance(df)
            print(f"Current Balance: ${balance:.2f}")

        elif choice == 4:
            save_data(df)
            print("Goodbye!")
            break
```
