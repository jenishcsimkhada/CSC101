## Problem 1: Tip Calculator  

### STEP 1: Understand the Problem  
This program helps restaurant staff in Perth calculate tips quickly and accurately. The user enters the bill amount and the tip percentage. The program then calculates the tip amount and the final total bill. This reduces manual errors and speeds up payment processing.

### STEP 2: Identify Inputs and Outputs  

**INPUTS:**
- Bill amount (e.g., $45.80) → numerical value in dollars  
- Tip percentage (e.g., 15%) → numerical percentage value  

**OUTPUTS:**
- Tip amount in dollars  
- Total bill amount including tip  
- Output displayed in currency format ($)


### STEP 3: Work It By Hand  

Bill amount = $45.80  
Tip percentage = 15%  

Convert percentage: 15% = 0.15  

Tip calculation:  
45.80 × 0.15 = 6.87  

Total calculation:  
45.80 + 6.87 = 52.67  

Final Answer:  
Tip = $6.87  
Total = $52.67  

## Problem 2: Late Fee Calculator  

### STEP 1: Understand the Problem  
This program is used in a Perth library to calculate overdue fines. The librarian enters how many days a book is late. The system calculates a fine of $0.50 per day. However, the maximum fine is capped at $10.00, meaning the user will never be charged more than this limit.

### STEP 2: Identify Inputs and Outputs  

**INPUT:**
- Number of overdue days (numerical value)

**OUTPUTS:**
- Fine amount in dollars  
- If the fine exceeds $10.00, it is capped at $10.00  
- Possible message: “Maximum fine reached”


### STEP 3: Work It By Hand  

Case A: 5 days late  
Calculation: 5 × 0.50 = 2.50  
Fine = $2.50  

Case B: 25 days late  
Calculation: 25 × 0.50 = 12.50  
Normal fine = $12.50  
But maximum allowed is $10.00  
Final fine = $10.00  

Special note: The fine is capped because it exceeds the maximum limit.

Case C: 0 days late  
Fine = $0.00
